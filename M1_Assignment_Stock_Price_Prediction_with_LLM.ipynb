```python
# Prompt: Give me the code that runs Load historical stock price data for Apple Inc. (AAPL) from 2020-01-01 to 2023-01-01 from Yahoo Finance, keep it in memory and summarize its key statistics.
import yfinance as yf
import pandas as pd

# Download historical data for AAPL from 2020-01-01 to 2023-01-01
data = yf.download("AAPL", start="2020-01-01", end="2023-01-01")
```

```python
# Summarize specific key statistics as requested
# Ensure 'Close' column exists and is numeric
if 'Close' in data.columns:
    highest_close = data['Close'].max()
    lowest_close = data['Close'].min()
    average_close = data['Close'].mean()
    total_trading_days = len(data)

    # Extract scalar values from Series for printing
    print(f"- **Highest Closing Price:** ${highest_close.iloc[0]:.2f}")
    print(f"- **Lowest Closing Price:** ${lowest_close.iloc[0]:.2f}")
    print(f"- **Average Closing Price:** ${average_close.iloc[0]:.2f}")
    print(f"- **Total Trading Days:** {total_trading_days}")
else:
    print("Close price data not found in the downloaded data.")
```

```python
#Prompt: Adopt the first four kinds of features and train a regression model to predict tomorrow's stock price. Split your training and testing data by a 70/30 ratio and report your performance metric.
!pip install ta

import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import ta
import pandas as pd

# Flatten MultiIndex columns if they exist. This is a common issue with yfinance data.
# Assuming columns are like ('Close', 'AAPL'), ('High', 'AAPL'), etc.
# We want to keep only the first level ('Close', 'High').
if isinstance(data.columns, pd.MultiIndex):
    data.columns = data.columns.droplevel(1)

# Ensure all necessary features are calculated
# Calculate Moving Averages
data['MA_5'] = data['Close'].rolling(window=5).mean()
data['MA_10'] = data['Close'].rolling(window=10).mean()
data['MA_20'] = data['Close'].rolling(window=20).mean()
data['MA_50'] = data['Close'].rolling(window=50).mean()

# Calculate Relative Strength Index (RSI)
data['RSI'] = ta.momentum.rsi(data['Close'], window=14)

# Calculate Moving Average Convergence Divergence (MACD)
data['MACD'] = ta.trend.macd(data['Close'])
data['MACD_Signal'] = ta.trend.macd_signal(data['Close'])
data['MACD_Diff'] = ta.trend.macd_diff(data['Close'])

# Create the target variable (tomorrow's closing price)
data['Tomorrow_Close'] = data['Close'].shift(-1)

# Define features (X) and target (y)
# First four kinds of features: Historical Prices, Volume, Moving Averages, Technical Indicators
features = [
    'Open', 'High', 'Low', 'Close', 'Volume',
    'MA_5', 'MA_10', 'MA_20', 'MA_50',
    'RSI', 'MACD', 'MACD_Signal', 'MACD_Diff'
]

X = data[features]
y = data['Tomorrow_Close']

# Drop rows with NaN values created by shifting and rolling window calculations
X = X.dropna()
y = y.dropna()

# Ensure X and y have the same number of samples after dropping NaNs
# This handles cases where dropping NaNs in X or y might result in different lengths
common_index = X.index.intersection(y.index)
X = X.loc[common_index]
y = y.loc[common_index]

# Split the data into training and testing sets (70/30 ratio)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Initialize and train the Linear Regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Report performance metrics
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"### Regression Model Performance Metrics:\n")
print(f"- **Mean Squared Error (MSE):** {mse:.2f}")
print(f"- **R-squared (R²) Score:** {r2:.3f}")

print("\nThese metrics indicate how well the model explains the variance in tomorrow's stock price. A lower MSE is better, and an R² closer to 1 indicates a better fit. A negative R² suggests the model performs worse than simply predicting the mean of the target variable.")
```

```python
# Prompt: Create a binary target variable based on historical prices (1 if the close price increase tomorrow, 0 other), and train a Random Forest model using the same train/test split ratio. Report feature importance and performance metrics.
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import pandas as pd

# Create the binary target variable
# 1 if tomorrow's close price is higher than today's close price, 0 otherwise
data['Tomorrow_Close_Direction'] = (data['Close'].shift(-1) > data['Close']).astype(int)

# Define features (X) and target (y)
# Using the same features as the regression model
features = [
    'Open', 'High', 'Low', 'Close', 'Volume',
    'MA_5', 'MA_10', 'MA_20', 'MA_50',
    'RSI', 'MACD', 'MACD_Signal', 'MACD_Diff'
]

X = data[features]
y = data['Tomorrow_Close_Direction']

# Drop rows with NaN values created by shifting and rolling window calculations
X = X.dropna()
y = y.dropna()

# Ensure X and y have the same number of samples after dropping NaNs
common_index = X.index.intersection(y.index)
X = X.loc[common_index]
y = y.loc[common_index]

# Split the data into training and testing sets (70/30 ratio)
# Using random_state for reproducibility
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Initialize and train the Random Forest Classifier model
model_rf = RandomForestClassifier(n_estimators=100, random_state=42)
model_rf.fit(X_train, y_train)

# Make predictions on the test set
y_pred_rf = model_rf.predict(X_test)

# Report performance metrics
accuracy_rf = accuracy_score(y_test, y_pred_rf)
conf_matrix_rf = confusion_matrix(y_test, y_pred_rf)
class_report_rf = classification_report(y_test, y_pred_rf)

print(f"### Random Forest Classifier Performance Metrics:\n")
print(f"- **Accuracy:** {accuracy_rf:.2f}")
print(f"- **Confusion Matrix:**\n{conf_matrix_rf}")
print(f"- **Classification Report:**\n{class_report_rf}")

# Report Feature Importance
feature_importance = pd.Series(model_rf.feature_importances_, index=features).sort_values(ascending=False)
print(f"\n### Feature Importance:\n{feature_importance}")
```
